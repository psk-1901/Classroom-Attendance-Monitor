```python
# Import necessary modules
import sys
from pathlib import Path

# Add project root to path
project_root = Path.cwd().parent
sys.path.insert(0, str(project_root))

from config import config
from database import db

print(f"Database Name: {config.DATABASE_NAME}")
print(f"MongoDB URI: {config.MONGODB_URI}")
```

```python
# Test database connection
try:
    stats = db.get_stats()

    print("✓ Database connection successful!")
    print("\nCollections:")

    for collection, count in stats.items():
        print(f"  {collection}: {count} documents")

except Exception as e:
    print(f"✗ Database connection failed: {e}")
```

```python
from pymongo import ASCENDING

def create_indexes():
    """Create indexes on collections."""

    # Students Collection
    students_col = db.get_collection(config.COLLECTIONS["students"])
    students_col.create_index([("roll_number", ASCENDING)], unique=True)
    students_col.create_index([("student_id", ASCENDING)])
    students_col.create_index([("email", ASCENDING)])
    students_col.create_index([("department", ASCENDING)])

    # Embeddings Collection
    embeddings_col = db.get_collection(config.COLLECTIONS["embeddings"])
    embeddings_col.create_index([("student_id", ASCENDING)], unique=True)
    embeddings_col.create_index([("roll_number", ASCENDING)])

    # Attendance Collection
    attendance_col = db.get_collection(config.COLLECTIONS["attendance"])
    attendance_col.create_index([("session_id", ASCENDING)])
    attendance_col.create_index([("student_id", ASCENDING)])
    attendance_col.create_index([("roll_number", ASCENDING)])
    attendance_col.create_index([("date", ASCENDING)])
    attendance_col.create_index([("department", ASCENDING)])
    attendance_col.create_index([("subject", ASCENDING)])

    # Logs Collection
    logs_col = db.get_collection(config.COLLECTIONS["logs"])
    logs_col.create_index([("timestamp", ASCENDING)])
    logs_col.create_index([("event_type", ASCENDING)])

    print("✓ Indexes created successfully")


create_indexes()
```
