```python
# Cell 4: View all registered students
all_students = student_manager.get_all_students()

print(f"\n📚 Total Students: {len(all_students)}")
print("\nStudent List:")
print("-" * 60)
print(f"{'Roll No':<10} | {'Name':<20} | {'Department':<20} | {'Section':<8}")
print("-" * 60)

for student in all_students:
    print(f"{student['roll_number']:<10} | {student['name']:<20} | {student['department']:<20} | {student['section']:<8}")
```

```python
# Cell 5: Search for a specific student
roll_number = input("\nEnter roll number to search: ").strip()

student = student_manager.get_student(roll_number)

if student:
    print("\n✅ Student Found:")
    print("-" * 40)
    print(f"Name: {student['name']}")
    print(f"Roll Number: {student['roll_number']}")
    print(f"Student ID: {student.get('student_id', 'N/A')}")
    print(f"Department: {student.get('department', 'N/A')}")
    print(f"Section: {student.get('section', 'N/A')}")
    print(f"Year: {student.get('year', 'N/A')}")
    print(f"Email: {student.get('email', 'N/A')}")
    print(f"Phone: {student.get('phone', 'N/A')}")
    print(f"Face Captured: {'✅ Yes' if student.get('face_captured', False) else '❌ No'}")
    print(f"Embeddings Generated: {'✅ Yes' if student.get('embeddings_generated', False) else '❌ No'}")
else:
    print(f"\n❌ Student with roll number {roll_number} not found")
```

```python
# Cell 6: Get student statistics
stats = student_manager.get_face_capture_status()

print("\n📊 Student Statistics:")
print("-" * 40)
print(f"Total Students: {stats['total_students']}")
print(f"Face Captured: {stats['face_captured']}")
print(f"Pending Capture: {stats['pending_capture']}")
print(f"Embeddings Generated: {stats['embeddings_generated']}")
print(f"Pending Embeddings: {stats['pending_embeddings']}")
```

```python
# Cell 7: Update student information
roll_number = input("\nEnter roll number to update: ").strip()

student = student_manager.get_student(roll_number)
if student:
    print(f"\nCurrent Details:")
    print(f"  Name: {student['name']}")
    print(f"  Department: {student['department']}")
    print(f"  Section: {student['section']}")
    print(f"  Year: {student['year']}")
    
    new_name = input("\nNew Name (press Enter to skip): ").strip()
    new_department = input("New Department (press Enter to skip): ").strip()
    new_section = input("New Section (press Enter to skip): ").strip()
    new_year = input("New Year (press Enter to skip): ").strip()
    
    update_data = {}
    if new_name:
        update_data['name'] = new_name
    if new_department:
        update_data['department'] = new_department
    if new_section:
        update_data['section'] = new_section
    if new_year:
        update_data['year'] = new_year
    
    if update_data:
        success, message = student_manager.update_student(roll_number, update_data)
        if success:
            print(f"\n✅ {message}")
        else:
            print(f"\n❌ {message}")
    else:
        print("\nℹ️ No updates made")
else:
    print(f"\n❌ Student {roll_number} not found")
```
