```python
# View captured images
from pathlib import Path

def show_captured_images(student_name, num_images=5):
    """Display captured images for a student."""

    student_dir = Path(f"../dataset/{student_name}")

    if not student_dir.exists():
        print(f"No images found for {student_name}")
        return

    images = sorted(student_dir.glob("*.jpg"))[:num_images]

    if not images:
        print(f"No images found for {student_name}")
        return

    fig, axes = plt.subplots(1, min(len(images), 5), figsize=(15, 3))

    for idx, img_path in enumerate(images[:5]):
        img = mpimg.imread(img_path)
        axes[idx].imshow(img)
        axes[idx].axis("off")
        axes[idx].set_title(f"Image {idx+1}")

    plt.suptitle(f"Captured Images: {student_name}")
    plt.tight_layout()
    plt.show()


student_name = input("\nEnter student name to view captured images: ").strip()

if student_name:
    show_captured_images(student_name)
```

```python
# Get dataset statistics
from pathlib import Path

dataset_dir = Path("../dataset")

if dataset_dir.exists():
    print("Dataset Statistics:")
    print("-" * 40)

    total_images = 0
    student_dirs = list(dataset_dir.iterdir())

    for student_dir in student_dirs:
        if student_dir.is_dir():
            images = list(student_dir.glob("*.jpg"))
            print(f"{student_dir.name}: {len(images)} images")
            total_images += len(images)

    print(f"\nTotal: {total_images} images across {len(student_dirs)} students")

else:
    print("Dataset directory not found")
```
